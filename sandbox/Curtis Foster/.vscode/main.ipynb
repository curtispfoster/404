```python
import pandas as pd
import matplotlib.pyplot as plt

# Fake sales data—replace with yours later
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [120, 150, 180, 210, 240]
}

df = pd.DataFrame(data)

print("Quick look at the data:")
print(df)

print("\nStats:")
print(df .describe())

# Plot it
plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales ($)")
plt.grid(True)
plt.show()
```
